Experiment 7: Encrypt the plaintext message using RSA algorithm. Then perform the reverse operation to get the original plaintext

```python
import random as r
from math import gcd, isqrt
```

```python
def is_prime(n):

    if n<2:

        return False

    for i in range(2,isqrt(n)+1):

        if n%i==0:

            return False

    return True
```

```python
is_prime(201)
```

```python
def gen_keys():

    primes=[i for i in range(1,50) if is_prime(i)]

    #print(len(primes))

    #print(primes)

    p=r.choice(primes)

    q=r.choice(primes)

    #return p,q

    if q==p:

        q=r.choice(primes)

    n=p*q

    phi=(p-1)*(q-1)

    e=r.randint(1,100000)

    #print(e)
    
    while gcd(phi,e)!=1 or e<=1 or e>=phi:
        e=r.randint(1,100000)

    d=pow(e,-1,phi)
    return (e,n),(d,n)
```

```python
def enc(m,e,n):

    return pow(m,e,n)
```

```python
def dec(c,d,n):

    return pow(c,d,n)
```

```python
public_key, private_key = gen_keys()
e, n = public_key
d, n = private_key

M = 42

if M >= n:
    print(f"M={M} > n={n}. Re-run.")
else:
    C = enc(M, e, n)
    D = dec(C, d, n)

    print(f"Public key  (e,n) : ({e},{n})")
    print(f"Private key (d,n) : ({d},{n})")
    print(f"Plaintext  M      : {M}")
    print(f"Ciphertext C      : {C}")
    print(f"Decrypted  M      : {D}")
```

```python
public_key, private_key = gen_keys()
e, n = public_key
d, n = private_key

M = "Hello"

if any(ord(ch) >= n for ch in M):
    print(f"n={n} is too small. Re-run.")
else:
    C = [enc(ord(ch), e, n) for ch in M]
    D = ''.join(chr(dec(c, d, n)) for c in C)

    print(f"Public key  (e,n) : ({e},{n})")
    print(f"Private key (d,n) : ({d},{n})")
    print(f"Plaintext  M      : {M}")
    print(f"Ciphertext C      : {C}")
    print(f"Decrypted  M      : {D}")
```
