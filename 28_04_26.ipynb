```python
import random

def lehmann_test(p, t=5):
    if p < 2:
        return False
    if p in (2, 3):
        return True
    if p % 2 == 0:
        return False

    for _ in range(t):
        a = random.randint(2, p - 2)
        r = pow(a, (p - 1) // 2, p)

        if r != 1 and r != p - 1:
            return False

    return True

P = int(input("Enter a number: "))

if lehmann_test(P):
    print("Probably Prime")
else:
    print("Composite")
```
