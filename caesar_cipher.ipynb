```python
def shift_text(text, shift):
    result = []  # Store converted characters

    for ch in text:
        if ch.isalpha():  # Check if character is a letter
            # Set base depending on uppercase or lowercase
            base = ord('A') if ch.isupper() else ord('a')

            # Apply shift with wrap-around using modulo
            new_char = chr((ord(ch) - base + shift) % 26 + base)

            result.append(new_char)  # Add shifted character
        else:
            result.append(ch)  # Keep numbers/symbols unchanged

    return "".join(result)  # Convert list to string

# Encryption function (shift +3)
def encrypt_message(message):
    return shift_text(message, 3)

# Decryption function (shift -3)
def decrypt_message(message):
    return shift_text(message, -3)

# Main program
text = "Shihab567"

# Encrypt the text
encrypted = encrypt_message(text)

# Decrypt the encrypted text
decrypted = decrypt_message(encrypted)

# Output results
print("Original Text :", text)
print("Encrypted Text:", encrypted)
print("Decrypted Text:", decrypted)
```
