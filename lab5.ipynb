```python
# T1: print primes
def sieve_print(n):
    # We assume every number from 0 to n could be prime at first
    sieve = [True] * (n + 1)
    
    # 0 and 1 are not prime, so we remove them right away
    sieve[0] = sieve[1] = False
    
    # We only need to check up to sqrt(n)
    for i in range(2, int(n**0.5) + 1):
        
        # If this number is still marked True, it's a prime
        if sieve[i]:
            
            # Now remove all multiples of this prime number
            for j in range(i * i, n + 1, i):
                sieve[j] = False
    
    # Finally, print all numbers that are still marked True (primes)
    for i, is_prime in enumerate(sieve):
        if is_prime:
            print(i)


# T2: return only primes array
def sieve_array(n):
    sieve = [True] * (n + 1)
    
    sieve[0] = sieve[1] = False
    
    for i in range(2, int(n**0.5) + 1):
        
        if sieve[i]:
            for j in range(i * i, n + 1, i):
                sieve[j] = False
    
    # Collect and return all numbers that are still True
    return [i for i, v in enumerate(sieve) if v]

# Print primes from 0 to 50 line by line
sieve_print(50)

# Get primes as a list
primes = sieve_array(50)

# Show the full list of primes
print(primes)
```
