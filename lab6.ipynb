```python
def to_lowercase(text):
    """
    This function just converts uppercase letters into lowercase.
    
    """

    result = ""

    for ch in text:
        # If it's uppercase, convert it
        if 'A' <= ch <= 'Z':
            result += chr(ord(ch) + 32)
        else:
            result += ch

    return result


# Example
msg = "HELLO Symmetric CRYPTO"
print(to_lowercase(msg))
```

```python
def encrypt(text, key):
    """
    This function encrypts text using a simple shift method.

    Each letter is shifted forward by the key value.
    """

    result = ""

    for ch in text:
        # Only handle lowercase letters
        if 'a' <= ch <= 'z':
            shifted = (ord(ch) - ord('a') + key) % 26
            result += chr(shifted + ord('a'))
        else:
            result += ch

    return result


def decrypt(text, key):
    """
    This function brings back the original text.

    It just shifts the letters backward using the same key.
    """

    result = ""

    for ch in text:
        if 'a' <= ch <= 'z':
            shifted = (ord(ch) - ord('a') - key) % 26
            result += chr(shifted + ord('a'))
        else:
            result += ch

    return result


# ---------------- Example Run ----------------

plain_text = "symmetric cipher"
key = 3

encrypted = encrypt(plain_text, key)
decrypted = decrypt(encrypted, key)

print("Original :", plain_text)
print("Encrypted:", encrypted)
print("Decrypted:", decrypted)
```
